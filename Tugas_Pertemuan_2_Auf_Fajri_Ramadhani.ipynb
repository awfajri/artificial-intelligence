2. Gunakan teknik pencarian Breadth - First search atau depth - first search

```python
# Representasi graph berdasarkan peta Romania
graph = {
    'Arad': ['Zerind', 'Sibiu', 'Timisoara'],
    'Zerind': ['Arad', 'Oradea'],
    'Oradea': ['Zerind', 'Sibiu'],
    'Sibiu': ['Arad', 'Oradea', 'Fagaras', 'Rimnicu Vilcea'],
    'Timisoara': ['Arad', 'Lugoj'],
    'Lugoj': ['Timisoara', 'Mehadia'],
    'Mehadia': ['Lugoj', 'Dobreta'],
    'Dobreta': ['Mehadia', 'Craiova'],
    'Craiova': ['Dobreta', 'Rimnicu Vilcea', 'Pitesti'],
    'Rimnicu Vilcea': ['Sibiu', 'Craiova', 'Pitesti'],
    'Fagaras': ['Sibiu', 'Bucharest'],
    'Pitesti': ['Rimnicu Vilcea', 'Craiova', 'Bucharest'],
    'Bucharest': ['Fagaras', 'Pitesti', 'Giurgiu', 'Urziceni'],
    'Giurgiu': ['Bucharest'],
    'Urziceni': ['Bucharest', 'Vaslui', 'Hirsova'],
    'Hirsova': ['Urziceni', 'Eforie'],
    'Eforie': ['Hirsova'],
    'Vaslui': ['Urziceni', 'Iasi'],
    'Iasi': ['Vaslui', 'Neamt'],
    'Neamt': ['Iasi']
}

def bfs_path(graph, start, goal):
    # Menyimpan path yang sedang dieksplorasi
    queue = [[start]]
    # Melacak node yang sudah dikunjungi
    visited = set()

    while queue:
        # Ambil path pertama dari antrean
        path = queue.pop(0)
        # Ambil node terakhir dari path tersebut
        node = path[-1]

        # Jika node terakhir adalah tujuan, kembalikan path
        if node == goal:
            return path

        # Jika node belum dikunjungi
        if node not in visited:
            # Eksplorasi tetangga
            for adjacent in graph.get(node, []):
                new_path = list(path)
                new_path.append(adjacent)
                queue.append(new_path)

            # Tandai node sebagai telah dikunjungi
            visited.add(node)

    return None

# Driver Code
start_city = 'Arad'
goal_city = 'Bucharest'
print("Pencarian Rute dengan BFS:")
print(bfs_path(graph, start_city, goal_city))
```

Pada kode Breadth-First Search (BFS) ini, algoritma bekerja menyisir rute secara merata lapis demi lapis (level by level) ke segala arah. Saya menggunakan list queue sebagai antrean untuk menyimpan rute yang sedang dicek, dan himpunan visited agar program tidak berputar-putar mengecek kota yang sama. Cara kerjanya mirip seperti menyebar luaskan pencarian ke semua kota tetangga terdekat secara bersamaan sebelum melangkah lebih dalam, sehingga ia pasti akan menemukan rute menuju Bucharest dengan cara mengeksplorasi semua kemungkinan secara sistematis.

3. Gunakan teknik pencarian Greedy Best First Search atau A*

```python
# Nilai heuristik (straight-line distance ke Bucharest)
h_table = {
    'Arad': 366,
    'Bucharest': 0,
    'Craiova': 160,
    'Dobreta': 242,
    'Eforie': 161,
    'Fagaras': 176,
    'Giurgiu': 77,
    'Hirsova': 151,
    'Iasi': 226,
    'Lugoj': 244,
    'Mehadia': 241,
    'Neamt': 234,
    'Oradea': 380,
    'Pitesti': 100,
    'Rimnicu Vilcea': 193,
    'Sibiu': 253,
    'Timisoara': 329,
    'Urziceni': 80,
    'Vaslui': 199,
    'Zerind': 374
}

# Representasi graph dengan jarak sebenarnya (cost)
graph_with_cost = {
    'Arad': [('Zerind', 75), ('Sibiu', 140), ('Timisoara', 118)],
    'Zerind': [('Arad', 75), ('Oradea', 71)],
    'Oradea': [('Zerind', 71), ('Sibiu', 151)],
    'Sibiu': [('Arad', 140), ('Oradea', 151), ('Fagaras', 99), ('Rimnicu Vilcea', 80)],
    'Timisoara': [('Arad', 118), ('Lugoj', 111)],
    'Lugoj': [('Timisoara', 111), ('Mehadia', 70)],
    'Mehadia': [('Lugoj', 70), ('Dobreta', 75)],
    'Dobreta': [('Mehadia', 75), ('Craiova', 120)],
    'Craiova': [('Dobreta', 120), ('Rimnicu Vilcea', 146), ('Pitesti', 138)],
    'Rimnicu Vilcea': [('Sibiu', 80), ('Craiova', 146), ('Pitesti', 97)],
    'Fagaras': [('Sibiu', 99), ('Bucharest', 211)],
    'Pitesti': [('Rimnicu Vilcea', 97), ('Craiova', 138), ('Bucharest', 101)],
    'Bucharest': [('Fagaras', 211), ('Pitesti', 101), ('Giurgiu', 90), ('Urziceni', 85)],
    'Giurgiu': [('Bucharest', 90)],
    'Urziceni': [('Bucharest', 85), ('Vaslui', 142), ('Hirsova', 98)],
    'Hirsova': [('Urziceni', 98), ('Eforie', 86)],
    'Eforie': [('Hirsova', 86)],
    'Vaslui': [('Urziceni', 142), ('Iasi', 92)],
    'Iasi': [('Vaslui', 92), ('Neamt', 87)],
    'Neamt': [('Iasi', 87)]
}

def path_h_cost(path):
    # Mengambil node terakhir dari path
    last_node = path[-1][0]
    # Mengambil nilai heuristik node terakhir
    h_cost = h_table[last_node]
    return h_cost

def greedy_best_search(graph, start, goal):
    visited = []
    # Antrean menyimpan path, bukan sekadar node
    queue = [[(start, 0)]]

    while queue:
        # Mengurutkan antrean berdasarkan nilai heuristik terendah (Greedy)
        queue.sort(key=path_h_cost)
        # Ambil path dengan h_cost terendah (yang ada di index pertama setelah di-sort)
        path = queue.pop(0)
        # Ambil node terakhir dari path tersebut
        node = path[-1][0]

        if node in visited:
            continue

        visited.append(node)

        if node == goal:
            return path
        else:
            adjacent_nodes = graph.get(node, [])
            for (node2, cost) in adjacent_nodes:
                new_path = path.copy()
                new_path.append((node2, cost))
                queue.append(new_path)

    return None

# Driver Code
start_city = 'Arad'
goal_city = 'Bucharest'
solution = greedy_best_search(graph_with_cost, start_city, goal_city)

print("\nPencarian Rute dengan Greedy Best-First Search:")
if solution:
    path_nodes = [node[0] for node in solution]
    print(path_nodes)
else:
    print("Rute tidak ditemukan")
```

Berbeda dengan BFS yang "buta arah" , pencarian menggunakan Greedy Best-First Search jauh lebih terarah karena memanfaatkan nilai heuristik (h_table), yaitu perkiraan jarak garis lurus dari setiap kota menuju Bucharest. Kunci utama di kode ini ada pada fungsi pengurutan (sorting) antrean; setiap kali program melihat opsi kota baru, ia akan langsung mengurutkan dan melangkah ke kota yang nilai heuristiknya paling kecil. Karena sifatnya yang "rakus" (Greedy) memprioritaskan kota yang kelihatannya paling dekat dengan tujuan akhir, proses pencariannya jadi jauh lebih cepat mencapai goal, meskipun total jarak tempuh aslinya belum tentu yang paling optimal

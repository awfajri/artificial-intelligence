1. inisialisasi sistem pakar

```python
# inisialisasi variabel
facts = []
is_changed = True

# fungsi untuk menambahkan fakta
def assert_fact(fact):
    global facts, is_changed

    if fact not in facts:
        facts.append(fact)
        is_changed = True
```

Pada bagian ini dibuat list facts yang digunakan untuk menyimpan seluruh fakta yang diketahui oleh sistem. Variabel is_changed digunakan untuk mengontrol proses looping pada forward chaining. Fungsi assert_fact berfungsi untuk menambahkan fakta baru dengan memastikan bahwa fakta tersebut belum ada sebelumnya sehingga tidak terjadi pengulangan.

2. Pemilihan Skenario oleh Pengguna

```python
# pilihan skenario
print("Pilih skenario pengujian:")
print("1. Cheetah")
print("2. Zebra")
print("3. Penguin")
print("4. Albatross")

pilihan = input("Masukkan pilihan (1-4): ")
obj = "hewan"

# menentukan fakta berdasarkan pilihan
if pilihan == "1":
    facts = [
        ["gives milk", obj],
        ["eats meat", obj],
        ["tawny color", obj],
        ["has dark spots", obj]
    ]

elif pilihan == "2":
    facts = [
        ["gives milk", obj],
        ["has hoofs", obj],
        ["has white color", obj],
        ["has black stripes", obj]
    ]

elif pilihan == "3":
    facts = [
        ["has feathers", obj],
        ["does not fly", obj],
        ["swims", obj],
        ["is black and white", obj]
    ]

elif pilihan == "4":
    facts = [
        ["has feathers", obj],
        ["is good flier", obj]
    ]

else:
    print("Pilihan tidak valid")

# tampilkan fakta awal sebelum inferensi
if pilihan in ["1", "2", "3", "4"]:
    print("\nFakta awal dari skenario:")
    for f in facts:
        print(f)
```

Pada bagian ini pengguna tidak perlu memasukkan fakta secara manual, melainkan cukup memilih skenario yang tersedia. Sistem kemudian mengisi fakta awal secara otomatis sesuai skenario yang dipilih. Fakta awal ditampilkan sebelum proses inferensi untuk menunjukkan kondisi awal yang akan diproses oleh sistem.

3. Proses Inferensi Forward Chaining

```python
# proses inferensi
while is_changed:
    is_changed = False
    for xf in facts:
        # Rule 1 - 10
        if xf[0] == "has hair":
            assert_fact(["is mammal", xf[1]])
        if xf[0] == "gives milk":
            assert_fact(["is mammal", xf[1]])
        if xf[0] == "has feathers":
            assert_fact(["is bird", xf[1]])
        if xf[0] == "flies" and ["lays eggs", xf[1]] in facts:
            assert_fact(["is bird", xf[1]])
        if xf[0] == "is mammal" and ["eats meat", xf[1]] in facts:
            assert_fact(["is carnivore", xf[1]])
        if xf[0] == "is mammal" and ["has pointed teeth", xf[1]] in facts and ["has claws", xf[1]] in facts and ["has forward pointing eyes", xf[1]] in facts:
            assert_fact(["is carnivore", xf[1]])
        if xf[0] == "is mammal" and ["has hoofs", xf[1]] in facts:
            assert_fact(["is ungulate", xf[1]])
        if xf[0] == "is mammal" and ["chews cud", xf[1]] in facts and ["is ungulate", xf[1]] in facts:
            assert_fact(["is even-toed", xf[1]])
        if xf[0] == "is carnivore" and ["tawny color", xf[1]] in facts and ["has dark spots", xf[1]] in facts:
            assert_fact(["is cheetah", xf[1]])
        if xf[0] == "is carnivore" and ["tawny color", xf[1]] in facts and ["has black stripes", xf[1]] in facts:
            assert_fact(["is tiger", xf[1]])

        # Rule 11 - 15
        if xf[0] == "is ungulate" and ["has long legs", xf[1]] in facts and ["has long neck", xf[1]] in facts and ["has tawny color", xf[1]] in facts and ["has dark spots", xf[1]] in facts:
            assert_fact(["is giraffe", xf[1]])

        if xf[0] == "is ungulate" and ["has white color", xf[1]] in facts and ["has black stripes", xf[1]] in facts:
            assert_fact(["is zebra", xf[1]])

        if xf[0] == "is bird" and ["does not fly", xf[1]] in facts and ["has long legs", xf[1]] in facts and ["has long neck", xf[1]] in facts and ["is black and white", xf[1]] in facts:
            assert_fact(["is ostrich", xf[1]])

        if xf[0] == "is bird" and ["does not fly", xf[1]] in facts and ["swims", xf[1]] in facts and ["is black and white", xf[1]] in facts:
            assert_fact(["is penguin", xf[1]])

        if xf[0] == "is bird" and ["is good flier", xf[1]] in facts:
            assert_fact(["is albatross", xf[1]])
```

Pada bagian ini sistem melakukan proses inferensi dengan mencocokkan fakta terhadap rule yang tersedia. Setiap rule yang terpenuhi akan menghasilkan fakta baru yang kemudian digunakan kembali dalam proses berikutnya hingga tidak ada fakta tambahan yang dihasilkan.

4. Evaluasi Hasil Inferensi

```python
# tampilkan seluruh hasil inferensi
print("\nSeluruh jejak fakta & inferensi:")
for f in facts:
    print(f)

# ambil kesimpulan akhir
kesimpulan = []
for f in facts:
    if f[0].startswith("is") and f[0] not in ["is mammal", "is bird", "is carnivore", "is ungulate", "is even-toed"]:
        kesimpulan.append(f)

# evaluasi hasil
print("\nEvaluasi:")
if len(kesimpulan) > 0:
    print("Sistem berhasil mengidentifikasi:")
    for k in kesimpulan:
        print(k)
else:
    print("Sistem belum berhasil mengidentifikasi hewan.")
```

Pada bagian ini sistem menampilkan seluruh jejak fakta, baik fakta awal maupun hasil inferensi. Setelah itu sistem menentukan kesimpulan akhir dengan menyaring fakta yang merupakan hasil identifikasi spesifik. Evaluasi dilakukan untuk memastikan apakah sistem berhasil mencapai kesimpulan berdasarkan rule yang tersedia.

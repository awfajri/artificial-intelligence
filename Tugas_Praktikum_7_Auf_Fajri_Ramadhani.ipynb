#Langkah 1: Persiapan Variabel dan Representasi Gen

```python
import random

POPULATION_SIZE = 20

GENES = [0, 1]

CHROMOSOME_LENGTH = 5
```

Langkah pertama, saya menyiapkan pengaturan dasar untuk algoritma genetikanya. Karena sekarang kita berhadapan dengan angka batas $0 \le x \le 31$, saya tidak bisa lagi menggunakan huruf sebagai gen. Sebagai gantinya, saya menggunakan sistem biner yaitu angka 0 dan 1. Untuk panjang kromosomnya, saya atur menjadi 5 bit. Alasannya simpel, angka maksimal yang diminta di soal adalah 31, dan kalau kita terjemahkan ke dalam bahasa komputer, angka 31 itu sama dengan 11111 yang terdiri dari 5 digit biner. Untuk ukuran populasinya cukup saya set di angka 20 agar proses komputasinya cepat.

#Langkah 2: Mendefinisikan Kelas Individu dan Fungsi Fitness

```python
class Individual:
    def __init__(self, chromosome):
        self.chromosome = chromosome
        # Menghitung nilai desimal (x) dari kromosom biner
        self.x_value = self.decode_chromosome()
        # Menghitung nilai fitness berdasarkan rumus soal
        self.fitness = self.cal_fitness()

    @classmethod
    def mutated_genes(cls):
        # Mengambil acak gen biner (0 atau 1)
        global GENES
        return random.choice(GENES)

    @classmethod
    def create_gnome(cls):
        # Membuat kromosom berisi 5 bit biner acak
        global CHROMOSOME_LENGTH
        return [cls.mutated_genes() for _ in range(CHROMOSOME_LENGTH)]

    def decode_chromosome(self):
        # Mengubah array biner (misal [1,0,0,1,1]) menjadi angka desimal
        binary_string = "".join(str(i) for i in self.chromosome)
        return int(binary_string, 2)

    def cal_fitness(self):
        # Rumus optimasi: f(x) = -(x^2) + 10
        return -(self.x_value ** 2) + 10

    def mate(self, par2):
        # Proses kawin silang (crossover) bawaan modul
        child_chromosome = []
        for gp1, gp2 in zip(self.chromosome, par2.chromosome):
            prob = random.random()
            if prob < 0.45:
                child_chromosome.append(gp1)
            elif prob < 0.90:
                child_chromosome.append(gp2)
            else:
                child_chromosome.append(self.mutated_genes())
        return Individual(child_chromosome)
```

Di langkah kedua ini, saya membuat kerangka untuk masing-masing individu (kromosom). Karena gennya berupa array biner, saya buatkan dulu fungsi decode_chromosome untuk mengubah susunan 0 dan 1 itu kembali menjadi angka desimal biasa (yaitu nilai $x$). Setelah nilai $x$ didapatkan, barulah angka tersebut saya masukkan ke dalam fungsi cal_fitness yang isinya adalah rumus dari soal yaitu $-(x^2) + 10$. Sisa fungsi lainnya seperti proses persilangan (mate) dan mutasi tidak saya ubah banyak dan tetap mempertahankan logika dari modul praktikum bab 7.  

#Langkah 3: Fungsi Utama dan Evolusi Populasi

```python
def main():
    global POPULATION_SIZE
    generation = 1
    found = False
    population = []

    # Membuat populasi awal
    for _ in range(POPULATION_SIZE):
        gnome = Individual.create_gnome()
        population.append(Individual(gnome))

    print("=== PROSES EVOLUSI ALGORITMA GENETIKA ===")
    while not found:
        # Urutkan secara DESCENDING (terbesar ke terkecil)
        # Karena kita mencari nilai Maksimum
        population = sorted(population, key=lambda x: x.fitness, reverse=True)

        # Kondisi Berhenti: Karena f(x) = -(x^2) + 10, nilai maksimal pasti 10 (saat x = 0)
        if population[0].fitness >= 10:
            found = True
            break

        new_generation = []

        # Elitism: Mempertahankan 10% individu terbaik ke generasi selanjutnya
        s = int((10 * POPULATION_SIZE) / 100)
        new_generation.extend(population[:s])

        # Kawin silang dari 50% populasi terbaik
        s = int((90 * POPULATION_SIZE) / 100)
        half_pop = int(POPULATION_SIZE / 2)
        for _ in range(s):
            parent1 = random.choice(population[:half_pop])
            parent2 = random.choice(population[:half_pop])
            child = parent1.mate(parent2)
            new_generation.append(child)

        population = new_generation

        # Menampilkan individu terbaik di tiap generasi
        print(f"Generasi: {generation}\tKromosom: {population[0].chromosome}\tNilai x: {population[0].x_value}\tFitness: {population[0].fitness}")
        generation += 1

    # Menampilkan hasil akhir
    print("\n=== SOLUSI OPTIMAL DITEMUKAN ===")
    print(f"Generasi ke-{generation}")
    print(f"Kromosom Terbaik : {population[0].chromosome}")
    print(f"Nilai x Optimal  : {population[0].x_value}")
    print(f"Nilai Maksimum f(x): {population[0].fitness}")

if __name__ == '__main__':
    main()
```

Pada langkah terakhir, saya menyusun alur utama evolusinya. Pada kode string asli, algoritma diurutkan dari yang terkecil karena mencari nilai error 0. Tapi karena soal ini meminta optimasi nilai maksimal, logika sorting-nya saya balik (reverse) dari yang paling besar ke yang paling kecil. Selain itu, saya memasang kondisi berhenti (break) ketika nilai fitness sudah mencapai 10. kenapa 10? Berdasarkan analisis logika matematika secara manual, fungsi maksimum dari $-(x^2) + 10$ pasti bernilai 10 (yaitu ketika AI berhasil menebak kromosom menjadi angka 0). Setelah dijalankan, kecerdasan buatan ini akan otomatis mengeliminasi tebakan yang jelek dan mencari kombinasi gen biner terbaik sampai ketemu solusinya.
